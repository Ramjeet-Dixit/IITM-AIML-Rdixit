```python
from smolagents import CodeAgent, tool

@tool
def calculator(expression: str) -> str:
    """
    Evaluates a basic math expression.

    Args:
        expression: A math expression as a string, e.g. "47 * 89 + 12"
    """
    try:
        return str(eval(expression))
    except Exception as e:
        return f"Error: {e}"

agent = CodeAgent(tools=[calculator], model=model)

result = agent.run("What is 47 * 89 + 12?")
print(result)
```

```python
@tool
def get_temperature(city: str) -> str:
    """
    Returns the current temperature in Celsius for a given city (mock data).

    Args:
        city: Name of the city, e.g. "London"
    """
    fake_data = {
        "london": 18,
        "delhi": 34,
        "new york": 22,
        "tokyo": 27,
    }
    temp = fake_data.get(city.lower())
    if temp is None:
        return f"No data for {city}"
    return f"{temp}°C"

agent = CodeAgent(tools=[calculator, get_temperature], model=model)

result = agent.run(
    "What is the temperature in Delhi in Celsius, and what would that be if you doubled it and added 10?"
)
print(result)
```
